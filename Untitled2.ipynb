```python
import random
import matplotlib.pyplot as plt

"This is a code that simulates data for x cancer cell lines."
"For each cell line, we create a random gene expression and random noise, which we then use to calculate the cell's drug response "
"The response depends on the gene expression, which we model with the parameters beta_0, beta_1 and sigma"
"Model drug response as DR[i,j] = beta_0[j] + beta_1[j] * G[i] + noise[i,j]. G is the gene expression, i is the cell line, j is the drug"

# number of cell lines: 479
n = 479

# number of drugs: 24
drugs = 24

# create "true" parameters for each drug
beta_0 = []
for j in range(drugs):
  beta_0.append(random.gauss(0, 0.5))

beta_1 =[]
for j in range(drugs):
  beta_1.append(random.gauss(0, 1))

# gene expression
mu_G = 0
sigma_G = 0.98

# noise
sigma_E = 0.54
mu_E = 0

# Create gene expression G for each gene
G = []

for _ in range(n):
  gene_expression = random.gauss(mu_G, sigma_G)
  G.append(gene_expression)

# Create noise
noise = []

for _ in range(n):
  cellnoise = []
  for j in range(drugs):
    error = random.gauss(mu_E, sigma_E)
    cellnoise.append(error)
  noise.append(cellnoise)

# Create drug response, DR
DR = []

for j in range(drugs):
  drug_responses = []
  for i in range(n):
    drugresponse = beta_0[j] + beta_1[j] * G[i] + noise[i][j]
    drug_responses.append(drugresponse)

  DR.append(drug_responses)

# Plot
plt.scatter(G, DR[0])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 1")
plt.show()

plt.scatter(G, DR[1])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 2")
plt.show()

plt.scatter(G, DR[2])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 3")
plt.show()
```

```python
" Model MLE and parameter estimation for drug 1"

# We pretend that we do not know the true beta_0 and beta_1 and estimate them from the simulated data

# Calculate the mean of gene expression
mean_G = sum(G) / len(G)

# Calculate the mean of drug response for Drug 1
mean_DR = sum(DR[0]) / len(DR[0])

# Estimate beta_1
numerator = 0
denominator = 0

for i in range(n):
    numerator = numerator + (G[i] - mean_G) * (DR[0][i] - mean_DR)
    denominator = denominator + (G[i] - mean_G) ** 2

beta_1_hat = numerator / denominator


# Estimate beta_0
beta_0_hat = mean_DR - beta_1_hat * mean_G


print("True beta_0:", beta_0[0])
print("Estimated beta_0:", beta_0_hat)

print("True beta_1:", beta_1[0])
print("Estimated beta_1:", beta_1_hat)
```

```python
import matplotlib.pyplot as plt
import random

# concentration range: 8 µM to 2.5 nM (8 point dose response assays) according to article, values inbetween were chosen arbitrarily
conc = [0.0025, 0.01, 0.04, 0.16, 0.64, 2.0, 4.0, 8.0] # µM
mu_E = 0
sigma_E = 0.01

# Förenkla Hill equation till 2 parametrar för att modellera läkemedelsrespons
def mu(c):
  EC50 = 1.5
  h = 2
  return 1 /(1 + (c / EC50)**h)

observations = []

for c in conc:
  expected_response = mu(c)
  E = random.gauss(mu_E, sigma_E)
  observation = mu(c) + E
  observations.append(observation)


  print("Concentration:", c, "µM")
  print("Expected response:", expected_response)
  print("Error:", E)
  print("Observed response:", observation)
  print()

plt.scatter(conc, observations)
plt.xlabel("Concentration (µM)")
plt.ylabel("Observed response")
plt.title("Drug concentration vs. observed response")
plt.show()
```

```python
import math

def nll(observations, theta):
  """NLL(theta) = - sum log p(y_i | theta)."""
  total = 0.0
  EC50, h = theta

  for i in range(len(observations)):
    x = observations[i]
    c = conc[i]

    mu = 1 /(1 + (c / EC50) ** h)

    log_p = (-0.5 * math.log(2 * math.pi * sigma_E**2) - (x - mu)**2 / (2 * sigma_E ** 2))

    total = total - log_p

  return total

theta = (1.5, 2)
result = nll(observations, theta)
print("NLL for the chosen parameters:", result)

EC50_hat = 0.0
h_hat = 0.0

best = nll(observations, (1.5, 2))

for a in range(201):
  EC50 = 0.5 + 0.01 * a

  for b in range(251):
    h = 0.5 + 0.01 * b

    score = nll(observations, (EC50, h))

    if score < best:
      best = score
      EC50_hat = EC50
      h_hat = h

print("Estimated EC50:", EC50_hat)
print("Estimated h:", h_hat)
print("NLL:", best)

```
