```python
import random
import matplotlib.pyplot as plt

"This is a code that simulates data for x cancer cell lines."
"For each cell line, we create a random gene expression and random noise, which we then use to calculate the cell's drug response "
"The response depends on the gene expression, which we model with the parameters beta_0, beta_1 and sigma"
"Model drug response as DR[i,j] = beta_0[j] + beta_1[j] * G[i] + noise[i,j]. G is the gene expression, i is the cell line, j is the drug"

# number of cell lines: 379. Trying the model on 50 cell lines
n = 50

# number of drugs: 24. Trying the model on 3 drugs
drugs = 3

# "True" parameters (choose these)
beta_0 = [0, 0, 0] # base level of response
beta_1 = [1.5, -0.2, 0.7]  # how strongly the gene expression affects the anticancer drug response

# gene expression
mu_G = 0
sigma_G = 0.98

# noise
sigma_E = 0.54
mu_E = 0

# Create gene expression G for each gene
G = []

for _ in range(n):
  gene_expression = random.gauss(mu_G, sigma_G)
  G.append(gene_expression)

# Create noise
noise = []

for _ in range(n):
  cellnoise = []
  for j in range(drugs):
    error = random.gauss(mu_E, sigma_E)
    cellnoise.append(error)
  noise.append(cellnoise)

# Create drug response, DR
DR = []

for j in range(drugs):
  drug_responses = []
  for i in range(n):
    drugresponse = beta_0[j] + beta_1[j] * G[i] + noise[i][j]
    drug_responses.append(drugresponse)

  DR.append(drug_responses)

# Plot
plt.scatter(G, DR[0])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 1")
plt.show()

plt.scatter(G, DR[1])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 2")
plt.show()

plt.scatter(G, DR[2])
plt.xlabel("Gene expression")
plt.ylabel("Anticancer drug response")
plt.title("Simulated cancer cell lines, drug 3")
plt.show()
```
